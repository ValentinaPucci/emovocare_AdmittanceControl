## Characterizing the flex sensor

At rest it reads ~11.57 kΩ with roughly ±40 Ω of noise (~0.3%), so any bend that changes resistance by more than a few percent is easy to resolve.

A single flex sensor gives one scalar, resistance. It can only tell you one geometric parameter, effectively the average curvature over the bent region. If you assume the sensor bends in a constant-curvature arc, everything else follows from that one number:

For a sensor of length L bent to radius r (base clamped, tangent along x):

Bend angle: θ = L / r
Tip position: x = r·sin θ, y = r·(1 − cos θ). The tip offset is y.
Tip angle: θ
Chord length: 2r·sin(θ/2)

Anything beyond that, like a localized kink versus a gentle arc, or the location of the bend, can’t be separated with one sensor. Real sensors also respond mostly where they bend, so the constant-curvature assumption is a model you should test, not a fact.
